## Structure

Refer to the [structure example](#example_structure) for more details.

In this example the sample {py:class}`Structure <udkm1Dsim.structures.structure.Structure>` consists of {py:class}`AmorphousLayer <udkm1Dsim.structures.layers.AmorphousLayer>`s.

In order to build an antiferromagnetic sample two different types of Fe {py:class}`Atom <udkm1Dsim.structures.atoms.Atom>`s and {py:class}`AmorphousLayer <udkm1Dsim.structures.layers.AmorphousLayer>`s are created with opposite in-plane magnetization.

Here a local file for the atomic scattering factor is read only for Fe atoms.

```python
Fe_right = ud.Atom(
    "Fe",
    magnetization={"r": 1, "phi": 90 * u.deg, "gamma": 90 * u.deg},
    atomic_form_factor_path="./Fe.cf",
)
Fe_left = ud.Atom(
    "Fe",
    magnetization={"r": 1, "phi": 90 * u.deg, "gamma": 270 * u.deg},
    atomic_form_factor_path="./Fe.cf",
)
Cr = ud.Atom("Cr")
Si = ud.Atom("Si")
```

```python
# both Fe layers share the same parameter groups
structural_Fe = ud.StructuralParameters(thickness=1 * u.nm, density=7874 * u.kg / u.m**3)
thermal_Fe = ud.ThermalParameters(
    heat_capacity=449 * u.J / u.kg / u.K,
    therm_cond=80 * u.W / (u.m * u.K),
    lin_therm_exp=11.8e-6,
)
elastic_Fe = ud.ElasticParameters(sound_vel=4.910 * u.nm / u.ps)
optical_Fe = ud.OpticalParameters(opt_ref_index=2.9174 + 3.3545j)

layer_Fe_left = ud.AmorphousLayer(
    "Fe_left",
    "Fe left amorphous",
    atom=Fe_left,
    structural=structural_Fe,
    thermal=thermal_Fe,
    elastic=elastic_Fe,
    optical=optical_Fe,
)
layer_Fe_right = ud.AmorphousLayer(
    "Fe_right",
    "Fe right amorphous",
    atom=Fe_right,
    structural=structural_Fe,
    thermal=thermal_Fe,
    elastic=elastic_Fe,
    optical=optical_Fe,
)
```

```python
layer_Cr = ud.AmorphousLayer(
    "Cr",
    "Cr amorphous",
    atom=Cr,
    structural=ud.StructuralParameters(thickness=1 * u.nm, density=7140 * u.kg / u.m**3),
    thermal=ud.ThermalParameters(
        heat_capacity=449 * u.J / u.kg / u.K,
        therm_cond=94 * u.W / (u.m * u.K),
        lin_therm_exp=6.2e-6,
    ),
    elastic=ud.ElasticParameters(sound_vel=5.940 * u.nm / u.ps),
    optical=ud.OpticalParameters(opt_ref_index=3.1612 + 3.4606j),
)
```

```python
layer_Si = ud.AmorphousLayer(
    "Si",
    "Si amorphous",
    atom=Si,
    structural=ud.StructuralParameters(thickness=1 * u.nm, density=2336 * u.kg / u.m**3),
    thermal=ud.ThermalParameters(
        heat_capacity=703 * u.J / u.kg / u.K,
        therm_cond=150 * u.W / (u.m * u.K),
        lin_therm_exp=2.6e-6,
    ),
    elastic=ud.ElasticParameters(sound_vel=8.433 * u.nm / u.ps),
    optical=ud.OpticalParameters(opt_ref_index=3.6941 + 0.0065435j),
)
```

```python
S = ud.Structure("Fe/Cr AFM Super Lattice")

# create a sub-structure
DL = ud.Structure("Two Fe/Cr Double Layers")
DL.add_sub_structure(layer_Fe_left, 1)
DL.add_sub_structure(layer_Cr, 1)
DL.add_sub_structure(layer_Fe_right, 1)
DL.add_sub_structure(layer_Cr, 1)

S.add_sub_structure(DL, 10)
S.add_sub_structure(layer_Fe_left, 10)
S.add_sub_structure(layer_Si, 200)

# add semi-infinite, static substrate
S.add_substrate(layer_Si)
```

## Magnetization

The {py:class}`Magnetization <udkm1Dsim.simulations.magnetization.Magnetization>` class is currently only an interface to allow the user for defining specific magnetization dynamics depending on the `strain_map` and `temp_map`.

Here the magnetization as function of temperature is used as a simplified model to alter the transient magnetization amplitude.

```python
def magnetization_bloch(x, Tc, M0, beta):
    m = M0 * ((1 - (x / Tc) ** (3 / 2)) * np.heaviside(Tc - x, 0.5)) ** beta
    return m


plt.figure()
temperatures = np.r_[10:1000:0.1]
plt.plot(temperatures, magnetization_bloch(temperatures, 770, 1, 0.75))
plt.xlabel("Temperature (K)")
plt.ylabel("Magneitzation")
plt.show()
```

The {py:meth}`calc_magnetization_map <udkm1Dsim.simulations.magnetization.Magnetization.calc_magnetization_map>` method must be overwritten to allow for calculating magnetization dynamics.

```python
class SimpleMagnetization(ud.Magnetization):
    def calc_magnetization_map(self, delays, **kwargs):
        temp_map = kwargs["temp_map"]
        T_c = kwargs["T_c"]
        magnetization_map = np.zeros([len(delays), self.S.get_number_of_layers(), 3])
        handles = self.S.get_layer_vectors()[2]

        for i, handle in enumerate(handles):
            if handle.id in ["Fe_left", "Fe_right"]:
                magnetization_map[:, i, 0] = magnetization_bloch(temp_map[:, i], T_c, 1, 0.75)
            _, phi, gamma = handle.magnetic.magnetization.polar_rad
            magnetization_map[:, i, 1] = phi
            magnetization_map[:, i, 2] = gamma

        return magnetization_map
```

```python
mag = SimpleMagnetization(S, True)
mag.save_data = False

magnetization_map = mag.get_magnetization_map(
    delays, strain_map=strain_map, temp_map=temp_map, T_c=770
)
```

```python
plt.figure()
plt.pcolormesh(
    distances.to("nm").magnitude,
    delays.to("ps").magnitude,
    magnetization_map[:, :, 0],
    shading="auto",
)
plt.title("Magnetization Amplitude Map")
plt.ylabel("Delay (ps)")
plt.xlabel("Distance (nm)")
plt.colorbar()
plt.show()
```

## Initialize dynamical magnetic X-ray simulations

The {py:class}`XrayDynMag <udkm1Dsim.simulations.xrays.XrayDynMag>` class requires a {py:class}`Structure <udkm1Dsim.structures.structure.Structure>` object and a boolean `force_recalc` in order overwrite previous simulation results.

```python
dyn_mag = ud.XrayDynMag(S, True)
dyn_mag.disp_messages = True
dyn_mag.save_data = False
```

## Homogeneous magnetic X-ray scattering

For the case of homogeneously strained/magnetized samples, the dynamical magnetic X-ray scattering simulations can be greatly simplified, which saves a lot of computational time.

### $q_z$-scan

The {py:class}`XrayDynMag <udkm1Dsim.simulations.xrays.XrayDynMag>` object requires an `energy` and scattering vector `qz` to run the simulations.  
Both parameters can be arrays and the resulting reflectivity has a first dimension for the photon energy and the a second for the scattering vector.

The resulting reflectivity is always calculated for the actual magnetization (`R_hom`) of the sample, as well as for opposite magnetization (`R_hom_phi`).

```python
dyn_mag.energy = np.r_[600, 708] * u.eV  # set two photon energies
dyn_mag.qz = np.r_[0.01:5:0.01] / u.nm  # qz range
# this is the actual calculation
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
```

In this example half-order antiferromagnetic Bragg peaks appear only at the resonance.

```python
plt.figure()
plt.semilogy(dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}")
plt.semilogy(dyn_mag.qz[1, :], R_hom[1, :], label=f"{dyn_mag.energy[1]}")
plt.ylabel("Reflectivity")
plt.xlabel(r"$q_z$ (1/nm)")
plt.legend()
plt.show()
```

### Post-Processing

Simple convolution of the results with an arbitrary function handle.

```python
FWHM = 0.01 / 1e-10  # Angstrom
sigma = FWHM / 2.3548

handle = lambda x: np.exp(-(((x) / sigma) ** 2) / 2)
y_conv = dyn_mag.conv_with_function(R_hom[1, :], dyn_mag._qz[1, :], handle)

plt.figure()
plt.semilogy(dyn_mag.qz[0, :], R_hom[1, :], label=f"{dyn_mag.energy[1]}")
plt.semilogy(dyn_mag.qz[0, :], y_conv, label=f"{dyn_mag.energy[1]} convoluted")
plt.ylabel("Reflectivity")
plt.xlabel(r"$q_z$ (1/nm)")
plt.legend()
plt.show()
```

### Energy- and $q_z$-scan

```python
dyn_mag.energy = np.r_[690:730:0.1] * u.eV  # set the energy range
dyn_mag.qz = np.r_[0.01:5:0.01] / u.nm  # qz range
# this is the actual calculation
R_hom_gallery, R_hom_phi_gallery, _, _ = dyn_mag.homogeneous_reflectivity()
```

```python
plt.figure(figsize=[6, 8])
plt.subplot(2, 1, 1)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_hom_gallery[0, :], label=np.round(dyn_mag.energy[0]))
plt.semilogy(
    dyn_mag.qz[0, :].to("1/nm"), R_hom_gallery[180, :], label=np.round(dyn_mag.energy[180])
)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_hom_gallery[-1, :], label=np.round(dyn_mag.energy[-1]))

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.legend()
plt.title("Dynamical Magnetic X-ray")

plt.subplot(2, 1, 2)
plt.pcolormesh(
    dyn_mag.qz[0, :].to("1/nm").magnitude,
    dyn_mag.energy.magnitude,
    np.log10(R_hom_gallery[:, :]),
    shading="auto",
)
plt.title("Dynamical Magnetic X-ray")
plt.ylabel("Energy (eV)")
plt.xlabel("$q_z$ (nm$^{-1}$)")

plt.tight_layout()
plt.show()
```

### Polarization dependence

The {py:class}`XrayDynMag <udkm1Dsim.simulations.xrays.XrayDynMag>` allows to set the incoming and outgoing polarization of the X-rays.

```python
dyn_mag.energy = np.r_[710] * u.eV  # set two photon energies
dyn_mag.qz = np.r_[0.01:5:0.01] / u.nm  # qz range

plt.figure()
dyn_mag.set_polarization(3, 0)
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
plt.semilogy(
    dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}: sigma -> unpolarized", alpha=0.5
)

dyn_mag.set_polarization(3, 3)
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
plt.semilogy(
    dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}: sigma -> sigma", alpha=0.5
)

dyn_mag.set_polarization(3, 4)
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
plt.semilogy(dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}: sigma -> pi", alpha=0.5)

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.title("Dynamical X-ray Polarization Dependence")
plt.legend()
plt.show()
```

## Inhomogeneous dynamical magnetic X-ray scattering

The {py:meth}`inhomogeneous_reflectivity <udkm1Dsim.simulations.xrays.XrayDynMag.inhomogeneous_reflectivity>` method allows to calculate the transient magnetic X-ray reflectivity according to a `strain_map` and/or `magnetization_map`.

```python
dyn_mag.energy = np.r_[708] * u.eV  # set the energy range
dyn_mag.qz = np.r_[1:3.5:0.01] / u.nm  # qz range
dyn_mag.set_polarization(3, 0)
```

```python
R_seq, R_seq_phi, _, _ = dyn_mag.inhomogeneous_reflectivity(
    strain_map=strain_map, magnetization_map=magnetization_map
)
```

```python
plt.figure(figsize=[6, 8])
plt.subplot(2, 1, 1)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq[0, 0, :], label=np.round(delays[0]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq[40, 0, :], label=np.round(delays[40]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq[-1, 0, :], label=np.round(delays[-1]))

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.legend()
plt.title("Dynamical Magnetic X-ray")

plt.subplot(2, 1, 2)
plt.pcolormesh(
    dyn_mag.qz[0, :].to("1/nm").magnitude,
    delays.to("ps").magnitude,
    np.log10(R_seq[:, 0, :]),
    shading="auto",
)

plt.ylabel("Delay (ps)")
plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.colorbar()
plt.show()
```

## Parallel dynamical X-ray scattering

Parallelization is fundamentally implemented but needs further improvements.
It works similar as with {py:class}`XrayDyn <udkm1Dsim.simulations.xrays.XrayDyn>`.

:::{important}
You need to install the **udkm1Dsim** with the `parallel` option which essentially add the Dask package to the requirements:

    > pip install udkm1Dsim[parallel]

You can also install/add Dask manually, e.g. via pip:

    > pip install dask

:::

Please refer to the [Dask documentation](https://docs.dask.org/en/latest/) for more details on parallel computing in Python.


```python
try:
    from dask.distributed import Client

    client = Client()
    R_seq_par, R_seq_phi_par, _, _ = dyn_mag.inhomogeneous_reflectivity(
        strain_map=strain_map,
        magnetization_map=magnetization_map,
        calc_type="parallel",
        dask_client=client,
    )
    client.close()
except ImportError:
    pass
```

```python
plt.figure(figsize=[6, 8])
plt.subplot(2, 1, 1)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq_par[0, 0, :], label=np.round(delays[0]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq_par[40, 0, :], label=np.round(delays[40]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq_par[-1, 0, :], label=np.round(delays[-1]))

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.legend()
plt.title("Parallel Dynamical Magnetic X-ray")

plt.subplot(2, 1, 2)
plt.pcolormesh(
    dyn_mag.qz[0, :].to("1/nm").magnitude,
    delays.to("ps").magnitude,
    np.log10(R_seq_par[:, 0, :]),
    shading="auto",
)

plt.ylabel("Delay (ps)")
plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.colorbar()
plt.show()
```
