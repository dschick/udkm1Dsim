## Structure

Refer to the [structure example](#example_structure) for more details.

In this example the sample {py:class}`Structure <udkm1Dsim.structures.structure.Structure>` consists of {py:class}`AmorphousLayer <udkm1Dsim.structures.layers.AmorphousLayer>`s.

In order to build an antiferromagnetic sample two different types of Fe {py:class}`Atom <udkm1Dsim.structures.atoms.Atom>`s and {py:class}`AmorphousLayer <udkm1Dsim.structures.layers.AmorphousLayer>`s are created with opposite in-plane magnetization.

Here a local file for the atomic scattering factor is read only for Fe atoms.

The heat is described by a two-temperature model of the electrons and phonons as in the [LLB example](#example_LLB), which requires two sub-systems for all layers.

```python
# opposite cartesian in-plane magnetizations along y
Fe_right = ud.Atom("Fe", magnetization=(0, 1, 0), atomic_form_factor_path="./Fe.cf")
Fe_left = ud.Atom("Fe", magnetization=(0, -1, 0), atomic_form_factor_path="./Fe.cf")
Cr = ud.Atom("Cr")
Si = ud.Atom("Si")
```

```python
# both Fe layers are built from the same parameter groups, which are copied for every layer
structural_Fe = ud.StructuralParameters(thickness=1 * u.nm, density=7874 * u.kg / u.m**3)
# two-temperature model of the electrons and phonons
thermal_Fe = ud.ThermalParameters(
    heat_capacity=["0.085*T", 450 * u.J / (u.kg * u.K)],
    therm_cond=[80 * u.W / (u.m * u.K), 5 * u.W / (u.m * u.K)],
    sub_system_coupling=["-7e17*(T_0-T_1)", "7e17*(T_0-T_1)"],
    lin_therm_exp=[0, 11.8e-6 / u.K],
    sub_system_labels=["e", "p"],
)
elastic_Fe = ud.ElasticParameters(sound_vel=4.910 * u.nm / u.ps)
optical_Fe = ud.OpticalParameters(dielectric=ud.Dispersion.from_ref_index(2.9174 + 3.3545j))
# mean-field parameters, the magnetization is set by the Fe atoms
magnetic_Fe = ud.MagneticParameters(eff_spin=2, curie_temp=1044 * u.K)

layer_Fe_left = ud.AmorphousLayer(
    "Fe_left",
    "Fe left amorphous",
    atom=Fe_left,
    structural=structural_Fe,
    thermal=thermal_Fe,
    elastic=elastic_Fe,
    optical=optical_Fe,
    magnetic=magnetic_Fe,
)
layer_Fe_right = ud.AmorphousLayer(
    "Fe_right",
    "Fe right amorphous",
    atom=Fe_right,
    structural=structural_Fe,
    thermal=thermal_Fe,
    elastic=elastic_Fe,
    optical=optical_Fe,
    magnetic=magnetic_Fe,
)
```

```python
layer_Cr = ud.AmorphousLayer(
    "Cr",
    "Cr amorphous",
    atom=Cr,
    structural=ud.StructuralParameters(thickness=1 * u.nm, density=7140 * u.kg / u.m**3),
    thermal=ud.ThermalParameters(
        heat_capacity=["0.03*T", 449 * u.J / (u.kg * u.K)],
        therm_cond=[90 * u.W / (u.m * u.K), 5 * u.W / (u.m * u.K)],
        sub_system_coupling=["-5e17*(T_0-T_1)", "5e17*(T_0-T_1)"],
        lin_therm_exp=[0, 6.2e-6 / u.K],
        sub_system_labels=["e", "p"],
    ),
    elastic=ud.ElasticParameters(sound_vel=5.940 * u.nm / u.ps),
    optical=ud.OpticalParameters(dielectric=ud.Dispersion.from_ref_index(3.1612 + 3.4606j)),
)
```

```python
# Si is not metallic
layer_Si = ud.AmorphousLayer(
    "Si",
    "Si amorphous",
    atom=Si,
    structural=ud.StructuralParameters(thickness=1 * u.nm, density=2336 * u.kg / u.m**3),
    thermal=ud.ThermalParameters(
        heat_capacity=[100 * u.J / (u.kg * u.K), 603 * u.J / (u.kg * u.K)],
        therm_cond=[0, 150 * u.W / (u.m * u.K)],
        sub_system_coupling=[0, 0],
        lin_therm_exp=[0, 2.6e-6 / u.K],
        sub_system_labels=["e", "p"],
    ),
    elastic=ud.ElasticParameters(sound_vel=8.433 * u.nm / u.ps),
    optical=ud.OpticalParameters(dielectric=ud.Dispersion.from_ref_index(3.6941 + 0.0065435j)),
)
```

The refractive indices above are only valid in the optical range, i.e. for the excitation at 800 nm.
In the X-ray range, the dielectric functions of the layers follow from the atomic and magnetic form factors of their atoms, which are set by {py:meth}`set_dielectric_from_atoms() <udkm1Dsim.structures.layers.Layer.set_dielectric_from_atoms>` with the refractive indices as fallback, see the [dielectric example](#example_dielectric):

```python
# the refractive indices above are kept as fallback outside of the form factors
for layer in [layer_Fe_left, layer_Fe_right, layer_Cr, layer_Si]:
    layer.set_dielectric_from_atoms(fallback=layer.dielectric.eps0)

layer_Fe_left.optical
```

```python
S = ud.Structure("Fe/Cr AFM Super Lattice")

# create a sub-structure
DL = ud.Structure("Two Fe/Cr Double Layers")
DL.add_sub_structure(layer_Fe_left, 1)
DL.add_sub_structure(layer_Cr, 1)
DL.add_sub_structure(layer_Fe_right, 1)
DL.add_sub_structure(layer_Cr, 1)

S.add_sub_structure(DL, 10)
S.add_sub_structure(layer_Fe_left, 10)
S.add_sub_structure(layer_Si, 200)

# add semi-infinite, static substrate
S.add_substrate(layer_Si)
```

## Magnetization

Refer to the [LLG example](#example_LLG) for more details.

Besides the {py:class}`LLG <udkm1Dsim.simulations.magnetization.LLG>` and {py:class}`LLB <udkm1Dsim.simulations.magnetization.LLB>`, the {py:class}`Magnetization <udkm1Dsim.simulations.magnetization.Magnetization>` class is an interface to allow the user for defining specific magnetization dynamics depending on the `strain_map` and `temp_map`.

Here, the magnetization amplitude follows the mean-field magnetization $m_\mathrm{eq}(T_e)$ of the electron temperature, i.e. the `temp_sub_system` 0, as a simplified model, see {py:meth}`get_mean_field_mag_map <udkm1Dsim.simulations.magnetization.Magnetization.get_mean_field_mag_map>`.
It is given by the `eff_spin` and `curie_temp` of the {py:class}`MagneticParameters <udkm1Dsim.structures.parameters.MagneticParameters>` of the Fe layers.

:::{note}
This *adiabatic limit* assumes that the magnetization follows the temperature instantaneously, otherwise the [LLB](#example_LLB) is required.
:::

The {py:meth}`calc_magnetization_map <udkm1Dsim.simulations.magnetization.Magnetization.calc_magnetization_map>` method must be overwritten to allow for calculating magnetization dynamics.
It returns the `magnetization_map` in cartesian coordinates $[m_x, m_y, m_z]$, which is here the mean-field magnetization times the cartesian `magnetization` of every layer.

:::{versionchanged} 2.6.1
The `magnetization_map` of the magnetization simulations and its input of the {py:class}`Elzo <udkm1Dsim.simulations.scattering.elzo.Elzo>` are given in cartesian instead of polar coordinates by default.
The former polar coordinates are still available by `polar=True` of {py:meth}`get_magnetization_map <udkm1Dsim.simulations.magnetization.Magnetization.get_magnetization_map>` and {py:meth}`inhomogeneous_reflectivity <udkm1Dsim.simulations.scattering.elzo.Elzo.inhomogeneous_reflectivity>`.
A `calc_magnetization_map` of a custom magnetization simulation must return cartesian maps.
:::

```python
class SimpleMagnetization(ud.Magnetization):
    def calc_magnetization_map(self, delays, **kwargs):
        # temperature of the temp_sub_system, here the electrons
        temps = self.check_temp_map(kwargs["temp_map"], len(delays), self.S.get_number_of_layers())
        # amplitude from the mean-field magnetization of the temperature
        mf_mag_map = self.get_mean_field_mag_map(temps)
        # cartesian direction of the magnetization of every layer
        handles = self.S.get_layer_vectors()[2]
        directions = np.array([handle.magnetization.magnitude for handle in handles])
        return mf_mag_map[:, :, np.newaxis] * directions[np.newaxis, :, :]
```

```python
mag = SimpleMagnetization(S, True)
mag.save_data = False
mag.disp_messages = True

temperatures = np.linspace(1, 1200, 600)
temp_map_mf = np.tile(temperatures[:, np.newaxis], (1, S.get_number_of_layers()))
mf_mag_map = mag.get_mean_field_mag_map(temp_map_mf)
select_Fe = S.get_all_positions_per_unique_layer()["Fe_left"]

plt.figure()
plt.plot(temperatures, mf_mag_map[:, select_Fe[0]])
plt.xlabel("Temperature (K)")
plt.ylabel("$m_\\mathrm{eq}$")
plt.title("Mean-Field Magnetization of Fe")
plt.show()
```

```python
magnetization_map = mag.get_magnetization_map(delays, strain_map=strain_map, temp_map=temp_map)
```

```python
plt.figure()
plt.pcolormesh(
    distances.to("nm").magnitude,
    delays.to("ps").magnitude,
    np.linalg.norm(magnetization_map, axis=-1),
    shading="auto",
)
plt.title("Magnetization Amplitude Map")
plt.ylabel("Delay (ps)")
plt.xlabel("Distance (nm)")
plt.colorbar()
plt.show()
```

## Homogeneous magnetic X-ray scattering

For the case of homogeneously strained/magnetized samples, the dynamical magnetic X-ray scattering simulations can be greatly simplified, which saves a lot of computational time.

### $q_z$-scan

The {py:class}`Elzo <udkm1Dsim.simulations.scattering.elzo.Elzo>` object requires an `energy` and scattering vector `qz` to run the simulations.  
Both parameters can be arrays and the resulting reflectivity has a first dimension for the photon energy and the a second for the scattering vector.

The resulting reflectivity is always calculated for the actual magnetization (`R_hom`) of the sample, as well as for opposite magnetization (`R_hom_phi`).

```python
dyn_mag.energy = np.r_[600, 708] * u.eV  # set two photon energies
dyn_mag.qz = np.r_[0.01:5:0.01] / u.nm  # qz range
# this is the actual calculation
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
```

In this example half-order antiferromagnetic Bragg peaks appear only at the resonance.

```python
plt.figure()
plt.semilogy(dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}")
plt.semilogy(dyn_mag.qz[1, :], R_hom[1, :], label=f"{dyn_mag.energy[1]}")
plt.ylabel("Reflectivity")
plt.xlabel(r"$q_z$ (1/nm)")
plt.legend()
plt.show()
```

### Post-Processing

Simple convolution of the results with an arbitrary function handle.

```python
FWHM = 0.01 / 1e-10  # Angstrom
sigma = FWHM / 2.3548

handle = lambda x: np.exp(-(((x) / sigma) ** 2) / 2)
y_conv = dyn_mag.conv_with_function(R_hom[1, :], dyn_mag.experiment.qz.magnitude[1, :], handle)

plt.figure()
plt.semilogy(dyn_mag.qz[0, :], R_hom[1, :], label=f"{dyn_mag.energy[1]}")
plt.semilogy(dyn_mag.qz[0, :], y_conv, label=f"{dyn_mag.energy[1]} convoluted")
plt.ylabel("Reflectivity")
plt.xlabel(r"$q_z$ (1/nm)")
plt.legend()
plt.show()
```

### Energy- and $q_z$-scan

```python
dyn_mag.energy = np.r_[690:730:0.1] * u.eV  # set the energy range
dyn_mag.qz = np.r_[0.01:5:0.01] / u.nm  # qz range
# this is the actual calculation
R_hom_gallery, R_hom_phi_gallery, _, _ = dyn_mag.homogeneous_reflectivity()
```

```python
plt.figure(figsize=[6, 8])
plt.subplot(2, 1, 1)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_hom_gallery[0, :], label=np.round(dyn_mag.energy[0]))
plt.semilogy(
    dyn_mag.qz[0, :].to("1/nm"), R_hom_gallery[180, :], label=np.round(dyn_mag.energy[180])
)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_hom_gallery[-1, :], label=np.round(dyn_mag.energy[-1]))

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.legend()
plt.title("Dynamical Magnetic X-ray")

plt.subplot(2, 1, 2)
plt.pcolormesh(
    dyn_mag.qz[0, :].to("1/nm").magnitude,
    dyn_mag.energy.magnitude,
    np.log10(R_hom_gallery[:, :]),
    shading="auto",
)
plt.title("Dynamical Magnetic X-ray")
plt.ylabel("Energy (eV)")
plt.xlabel("$q_z$ (nm$^{-1}$)")

plt.tight_layout()
plt.show()
```

### Polarization dependence

The {py:class}`Elzo <udkm1Dsim.simulations.scattering.elzo.Elzo>` allows to set the incoming and outgoing polarization of the X-rays.

The polarizations are given by their names or their ids:

| id | name | incoming | analyzer |
| :--- | :--- | :--- | :--- |
| 0 | `"unpolarized"` | unpolarized | no analyzer, also `"none"` |
| 1 | `"circ +"` | circular + | circular + |
| 2 | `"circ -"` | circular - | circular - |
| 3 | `"sigma"` | $\sigma$ | $\sigma$ |
| 4 | `"pi"` | $\pi$ | $\pi$ |
| 5 | `"elliptical"` | elliptical | elliptical |

States which are not supported by a formalism raise a `ValueError`.


```python
dyn_mag.energy = np.r_[710] * u.eV  # set two photon energies
dyn_mag.qz = np.r_[0.01:5:0.01] / u.nm  # qz range

plt.figure()
dyn_mag.set_polarization("sigma", "none")
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
plt.semilogy(
    dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}: sigma -> unpolarized", alpha=0.5
)

dyn_mag.set_polarization("sigma", "sigma")
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
plt.semilogy(
    dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}: sigma -> sigma", alpha=0.5
)

dyn_mag.set_polarization("sigma", "pi")
R_hom, R_hom_phi, _, _ = dyn_mag.homogeneous_reflectivity()
plt.semilogy(dyn_mag.qz[0, :], R_hom[0, :], label=f"{dyn_mag.energy[0]}: sigma -> pi", alpha=0.5)

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.title("Dynamical X-ray Polarization Dependence")
plt.legend()
plt.show()
```

## Inhomogeneous dynamical magnetic X-ray scattering

The {py:meth}`inhomogeneous_reflectivity <udkm1Dsim.simulations.scattering.elzo.Elzo.inhomogeneous_reflectivity>` method allows to calculate the transient magnetic X-ray reflectivity according to a `strain_map` and/or `magnetization_map`.

```python
dyn_mag.energy = np.r_[708] * u.eV  # set the energy range
dyn_mag.qz = np.r_[1:3.5:0.01] / u.nm  # qz range
dyn_mag.set_polarization("sigma", "none")
```

```python
R_seq, R_seq_phi, _, _ = dyn_mag.inhomogeneous_reflectivity(
    strain_map=strain_map, magnetization_map=magnetization_map
)
```

```python
plt.figure(figsize=[6, 8])
plt.subplot(2, 1, 1)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq[0, 0, :], label=np.round(delays[0]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq[40, 0, :], label=np.round(delays[40]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq[-1, 0, :], label=np.round(delays[-1]))

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.legend()
plt.title("Dynamical Magnetic X-ray")

plt.subplot(2, 1, 2)
plt.pcolormesh(
    dyn_mag.qz[0, :].to("1/nm").magnitude,
    delays.to("ps").magnitude,
    np.log10(R_seq[:, 0, :]),
    shading="auto",
)

plt.ylabel("Delay (ps)")
plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.colorbar()
plt.show()
```

## Parallel dynamical X-ray scattering

Parallelization is fundamentally implemented but needs further improvements.
It works similar as with {py:class}`XrayDyn <udkm1Dsim.simulations.scattering.dynamical.XrayDyn>`.

:::{important}
You need to install the **udkm1Dsim** with the `parallel` option which essentially add the Dask package to the requirements:

    > pip install udkm1Dsim[parallel]

You can also install/add Dask manually, e.g. via pip:

    > pip install dask

:::

Please refer to the [Dask documentation](https://docs.dask.org/en/latest/) for more details on parallel computing in Python.


```python
try:
    from dask.distributed import Client

    client = Client()
    R_seq_par, R_seq_phi_par, _, _ = dyn_mag.inhomogeneous_reflectivity(
        strain_map=strain_map,
        magnetization_map=magnetization_map,
        calc_type="parallel",
        dask_client=client,
    )
    client.close()
except ImportError:
    pass
```

```python
plt.figure(figsize=[6, 8])
plt.subplot(2, 1, 1)
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq_par[0, 0, :], label=np.round(delays[0]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq_par[40, 0, :], label=np.round(delays[40]))
plt.semilogy(dyn_mag.qz[0, :].to("1/nm"), R_seq_par[-1, 0, :], label=np.round(delays[-1]))

plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.ylabel("Reflectivity")
plt.legend()
plt.title("Parallel Dynamical Magnetic X-ray")

plt.subplot(2, 1, 2)
plt.pcolormesh(
    dyn_mag.qz[0, :].to("1/nm").magnitude,
    delays.to("ps").magnitude,
    np.log10(R_seq_par[:, 0, :]),
    shading="auto",
)

plt.ylabel("Delay (ps)")
plt.xlabel("$q_z$ (nm$^{-1}$)")
plt.colorbar()
plt.show()
```
